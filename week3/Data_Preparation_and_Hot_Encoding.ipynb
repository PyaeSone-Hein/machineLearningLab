# #2. HANDLING MISSING DATA (MISSING VALUES)

```python
# check dataFrame
hr_df
```

```python
# Let’s find the rows that have Null values
# Notice row index 6 has many missing values
hr_df.isnull()
```

```python
# Let’s look at the number of missing values per column
hr_df.isnull().sum()
```

```python
# Drop all rows that contain Null values
# Note that the data size has been reduced by 7 elements from 1470 to 1463
hr_df.dropna(how='any', inplace=True)
# how='any' -> if any column in a row has a missing value, drop the entire row.
# alternative way: how='all' would only drop a row if all columns are missing
# inplace=True -> the change is applied immediately to the dataframe
```

```python
# Note that row 1 is no longer there!
hr_df.head()
```

```python
# Check if we still have missing values
hr_df.isnull().sum()
```

```python
# Explore alternative (smarter) methods to handle missing values
# Reload the raw data using Pandas as follows
hr_df = pd.read_csv('Human_Resources.csv')
hr_df
```

```python
# Calculate the average monthly income
hr_df['MonthlyIncome'].mean()
```

```python
# Use Fillna to fill a specific column with a specific value
hr_df.fillna({"MonthlyIncome": hr_df["MonthlyIncome"].mean()}, inplace=True)
```

# **ONE-HOT ENCODING**

Machine learning models generally require **numeric inputs**; variables stored as objects or strings must be encoded. Two standard approaches are **One-Hot (Dummy) Encoding** and **Ordinal (Label) Encoding**.

* **One-Hot Encoding** is preferred for **nominal** categories (no natural order), because it creates separate 0/1 indicator columns and does not impose an artificial ranking among categories.

* **Ordinal (Label)** Encoding should be used only when the categories have a meaningful order (e.g., Low < Medium < High). Assigning integers to unordered labels can mislead many models by implying nonexistent magnitude or distance.

image.png

For **binary** variables, use a **single 0/1** indicator (you can drop one dummy column). For example, for gender with values male and female, encode **Male = 1** if male, **0 otherwise**. This avoids redundant columns and potential multicollinearity in linear models.

image.png

```python
hr_df = pd.read_csv('Human_Resources.csv')
hr_df.head()
```
