```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
#Linear Regression Model
```

```python
data = pd.DataFrame({
    'Area (m2)': [50, 60, 70, 80, 90, 100], #feature
    'Price (million)': [150, 180, 200, 220, 240, 270] #target
})
```

```python
X = data[['Area (m2)']]
y = data[['Price (million)']]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
#80 percent train and 20 percent test
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

```python
#visualization
```

```python
plt.scatter(X_test, y_test, color = 'blue', label = 'Actual Data')
plt.plot(X_test, y_pred, color = 'red', linewidth = 2, label = 'Predicted Regression Line')
plt.xlabel('Area m2')
plt.ylabel('Price (million)')
plt.title('Linear Regression')
plt.legend()
plt.show()
```

```python
#Support Vector Regression(SVR)
```

```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
```

```python
#feature scaling(requires)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = SVR(kernel='rbf', C = 100, epsilon = 10)
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)
```

```python
print("SVR - MAE:", mean_absolute_error(y_test, y_pred))
print("SVR - MSE:", mean_squared_error(y_test, y_pred))
print("SVR - R² Score:", r2_score(y_test, y_pred))
```
