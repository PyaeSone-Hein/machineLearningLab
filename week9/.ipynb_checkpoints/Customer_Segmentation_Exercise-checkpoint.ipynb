```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

# **Import Data**

source: https://archive.ics.uci.edu/dataset/292/wholesale+customers

```python
df = pd.read_csv("customer.csv")
df.head()
```

```python
data_to_cluster = df.drop(['Channel', 'Region'], axis=1)
data_log_transformed = data_to_cluster.apply(np.log1p)
```

```python
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_log_transformed)
```

# **Exploratory Data Analysis**

```python
spending_cols = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']
```

```python
plt.figure(figsize=(18, 10))
plt.suptitle('Histograms of Spending Categories', fontsize=16)
for i, col in enumerate(spending_cols):
    plt.subplot(2, 3, i+1)
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(col)
    plt.xlabel('Amount')
    plt.ylabel('Frequency')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

# **Determine the number of clusters, training fit, predict, and cluster analysis**

```python
kmeans = KMeans(n_clusters=2) # check k=2 or k=3 or k=3 and see the result.
                              # do the same to X3
member = kmeans.fit_predict(data_scaled)
centre = kmeans.cluster_centers_

plt.figure(figsize=(6, 6))
plt.scatter(data_scaled[:, 0], data_scaled[:, 1], c=member, s=10) #this with color
plt.scatter(centre[:, 0], centre[:, 1], c="w", edgecolors="k", s=50, linewidths=2)
plt.axis("equal");
```

```python
score = []
for k in range(2, 25):
    kmeans = KMeans(n_clusters=k)
    member = kmeans.fit_predict(data_scaled)
    score.append(kmeans.inertia_)

plt.figure(figsize=(10, 5))
plt.plot( range(2, 25), score, "r-")
plt.ylabel("Inertia")
# the goal of the K-Means algorithm is to minimize inertia.
# inertia is calculated as the sum of the squared distances from each data point
# to its assigned cluster's centroid (center point).
# it is also very commonly called the Within-Cluster Sum of Squares (WCSS)
plt.xlabel("Number of Clusters");
```

```python
score = []
for k in range(2, 25):
    kmeans = KMeans(n_clusters=k)
    member = kmeans.fit_predict(data_scaled)
    silhouette = silhouette_score(data_scaled, member, metric='euclidean')
    score.append(silhouette)

plt.figure(figsize=(10, 5))
plt.plot( range(2, 25), score, "r-")
plt.ylabel("Silhouette")
plt.xlabel("Number of Clusters");
```
